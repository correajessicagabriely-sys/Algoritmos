### Exercicio 11 - Reajuste de salário

```python
salario = float(input("Digite o salário atual: R$ "))
if salario <= 280:
  perc = 20
elif salario <= 700:
  perc = 15
elif salario <= 1500:
  perc = 10
else:
  perc = 5

aumento = salario * (perc / 100)
novo_salario = salario + aumento
print(f"Salário antes do reajuste: R$ {salario:.2f}")
print(f"Percentual de aumento aplicado: {perc}%")
print(f"Valor do aumento: R$ {aumento:.2f}")
print(f"Novo salário: R$ {novo_salario:.2f}")
```

### Exercício 12 - Folha de pagamento

```python
valor_hora = float(input("Valor da hora: R$ "))
horas = float(input("Horas trabalhadas no mês: "))
salario_bruto = valor_hora * horas

if salario_bruto <= 900:
  ir_perc = 0
elif salario_bruto <= 1500:
  ir_perc = 5
elif salario_bruto <= 2500:
  ir_perc = 10
else:
  ir_perc = 20

ir = salario_bruto * (ir_perc / 100)
inss = salario_bruto * 0.10
fgts = salario_bruto * 0.11
total_descontos = ir + inss
salario_liquido = salario_bruto - total_descontos

print(f"Salário Bruto: R$ {salario_bruto:.2f}")
print(f"(-) IR ({ir_perc}%): R$ {ir:.2f}")
print(f"(-) INSS (10%): R$ {inss:.2f}")
print(f"FGTS (11%): R$ {fgts:.2f}")
print(f"Total de descontos: R$ {total_descontos:.2f}")
print(f"Salário Líquido: R$ {salario_liquido:.2f}")
```

### Exercicio 14 - Média de aproveitamento

### Exercicio 19 - Casas decimais

```python
num = int(input("Digite um número inteiro menor que 1000: "))
if num < 1000:
  c = num // 100
  d = (num % 100) // 10
  u = num % 10
  res = []
  if c > 0:
    res.append(f"{c} centena{'s' if c > 1 else ''}")
  if d > 0:
    res.append(f"{d} dezena{'s' if d > 1 else ''}")
  if u > 0:
    res.append(f"{u} unidade{'s' if u > 1 else ''}")

  if len(res) == 3:
    print(f"{res[0]}, {res[1]} e {res[2]}")
  elif len(res) == 2:
    print(f"{res[0]} e {res[1]}")
  elif len(res) == 1:
    print(res[0])
else:
  print("Número inválido.")
```

### Exercicio 23 - Operações numericas

```python
n1 = float(input("Número 1: "))
n2 = float(input("Número 2: "))
op = input("Escolha a operação (+, -, *, /): ")

if op == "+":
  res = n1 + n2
elif op == "-":
  res = n1 - n2
elif op == "*":
  res = n1 * n2
elif op == "/":
  res = n1 / n2 if n2 != 0 else 0
else:
  res = None

if res is not None:
  print(f"Resultado: {res}")
  print("Par" if res % 2 == 0 else "Ímpar")
  print("Positivo" if res > 0 else ("Negativo" if res < 0 else "Zero"))
  print("Inteiro" if res % 1 == 0 else "Decimal")
```

### Exercicio 24 - Investigação forense

```python
p1 = input("Telefonou para a vítima? (sim/nao): ")
p2 = input("Esteve no local do crime? (sim/nao): ")
p3 = input("Mora perto da vítima? (sim/nao): ")
p4 = input("Devia para a vítima? (sim/nao): ")
p5 = input("Já trabalhou com a vítima? (sim/nao): ")

sims = [p1, p2, p3, p4, p5].count("sim")
if sims == 5:
  print("Assassino")
elif 3 <= sims <= 4:
  print("Cúmplice")
elif sims == 2:
  print("Suspeita")
else:
  print("Inocente")
```

### Exercicio 25 - Posto de combustivel

```python
litros = float(input("Quantidade de litros: "))
tipo = input("Tipo (A-Álcool, G-Gasolina): ").upper()

if tipo == "A":
  preco = 1.90
  desc = 0.03 if litros <= 20 else 0.05
elif tipo == "G":
  preco = 2.50
  desc = 0.04 if litros <= 20 else 0.06
else:
  preco, desc = 0, 0

total = litros * preco * (1 - desc)
print(f"Valor a pagar: R$ {total:.2f}")
```

### Exercicio 35 - Intervalo em Sequencia

```python
# Exercício 35
inicio = int(input("Digite o primeiro número inteiro: "))
fim = int(input("Digite o segundo número inteiro: "))

if inicio > fim:
  inicio, fim = fim, inicio

print(f"Sequência entre {inicio} e {fim}:")
for i in range(inicio + 1, fim):
  print(i)
```

### Exercicio 39 - Quantidade de pares e impares

```python
# Exercício 39
pares = 0
impares = 0
for i in range(1, 11):
  n = int(input(f"Digite o {i}º número inteiro: "))
  if n % 2 == 0:
    pares += 1
  else:
    impares += 1

print(f"Total de números pares: {pares}")
print(f"Total de números ímpares: {impares}")
```
